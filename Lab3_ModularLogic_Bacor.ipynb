```python
# [CELL 4] PARAMETRIC PROCESSING
def user_summary(title, *scores, **info):
    print(f"=== {title} ===")

    total = sum(scores)
    average = total / len(scores)

    print("Scores:", scores)
    print ("Total:", total)
    print("Average:", round(average, 2))

    for k, v in info.items():
        print(f"{k}: {v}")

    return average

avg = user_summary(
    f"{LAST_NAME} Academic Report",
    SEED_DIGIT * 10,
    ID_SUM % 100,
    NAME_LENGTH * 5,
    id=STUDENT_ID,
    surname=LAST_NAME
)

```

```python
# [CELL 5] RETURN MECHANISM

def compute_area(radius):
    return 3.1416 * radius**2

area = compute_area(SEED_DIGIT + 2)
print("Computed Area:", round(area,2))
```

```python
# [CELL 6] SAFE CALCULATOR

def safe_calculator(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        result = "UNDEFINED"
    return result

print("Calculation Result:", safe_calculator(ID_SUM, SEED_DIGIT))
```

```python
# [CELL 7] RECURSION

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)
print("Factorial Result:", factorial(SEED_DIGIT))
```

```python
# In Cell 8, there are conflicting variables. 
# In the procedures, it is said to reverse LAST_NAME (Bacor)
# However, in Data and Observation, it is said to reverse STUDENT_ID (TUPM-26-1879)
# I will proceed to follow the Procedures.

# [CELL 8] STRING RECURSION

def reverse(text):
    if len(text) <= 1:
        return text
    return text[-1] + reverse(text[:-1])
print("Reversed Surname:", reverse(LAST_NAME.upper()))
```

```python
# [CELL 9] DECORATOR

def logger(func):
    def wrapper(*args, **kwargs):
        print("Executing:", func.__name__)
        return func(*args, **kwargs)
    return wrapper

@logger
def multiply(x, y):
    return x * y
print("Multiplication:", multiply(SEED_DIGIT, NAME_LENGTH))
```

```python
# [CELL 10] LAMBDA
data = list(range(1, SEED_DIGIT + 10))
squared = list(map(lambda x: x**2, data))

print("Squared Data:", squared)
```

```python
# [CELL 11] GENERATOR

def even_squares(limit):
    for i in range(limit):
        if i % 2 == 0:
            yield i**2

for value in even_squares(NAME_LENGTH + SEED_DIGIT):
    print(value)
```
