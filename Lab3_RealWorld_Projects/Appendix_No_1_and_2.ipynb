```python
# BELOW ARE APPENDIX 1 AND 2

## APPENDIX 1: EQUIPMENT DIAGNOSTIC SYSTEM

### DEFINING OF VARIABLES
LAST_NAME = "Bacor"
STUDENT_ID = "TUPM-26-1879"
SEED_NUM = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
FAVORITE_ARTIST = "IVOS"
ARTIST_LENGTH = len(FAVORITE_ARTIST)

# Set the limit for operating readings.
limits = (NAME_LENGTH + SEED_NUM + ARTIST_LENGTH) * 10 # Limit: 180


# Define Student-specific readings
# Decide valid, and invalid inputs 
readings = [
    NAME_LENGTH * 5, 
    SEED_NUM % 10, 
    "WRONG TYPE", # Wrong data type (str)
    ARTIST_LENGTH * -1, # Negative value
    NAME_LENGTH + 15, 
    ARTIST_LENGTH + NAME_LENGTH * 8,
    (SEED_NUM + NAME_LENGTH + ARTIST_LENGTH) * 12 # Out of Bounds
]

def diagnostic_monitor(func):
    def wrapper(*args, **kwargs):
        print(f"Monitoring... Currently Executing: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Done Monitoring {func.__name__}")
        return result
    return wrapper

@diagnostic_monitor
def validate_data(*input):
    valid_data = []
    invalid_data = []
    for val in input:
        try:
            # Ensure every input is integer.
            val_int = int(val)

            # Filtering usual invalid input
            ## Out of Bounds
            if val_int > limits:
                invalid_data.append(val)
            ## Negative
            elif val_int < 0:
                invalid_data.append(val)
            ## Low Signal
            elif val_int < 5:
                invalid_data.append(val)
            else:
                valid_data.append(val)
        except (ValueError, TypeError):
            # Filtering rare/unexpected invalid inputs.
            invalid_data.append(val)
    return valid_data, invalid_data

@diagnostic_monitor
def calculate_readings(valid_data, invalid_data):
    average_valid = len(valid_data)
    average_invalid = len(invalid_data)
    return average_valid, average_invalid

@diagnostic_monitor
def classify_results(ave_valid, ave_invalid):
    classification_rate = round((ave_invalid / ave_valid), 2)
    if classification_rate <= 0.25:
        return "Optimal Condition."
    elif 0.30 < classification_rate <= 0.75:
        return "WARNING. ELEVATED THREAT."
    else:
        return "CRITICAL CONDITION. PLEASE CHECK THE SYSTEM IMMEDIATELY."

valid, invalid = validate_data(*readings)    
average_valid, average_invalid = calculate_readings(valid_data=valid, invalid_data=invalid)
result = classify_results(ave_valid=average_valid, ave_invalid=average_invalid)

print("=" * 40)
print(f"VALIDATION RESULTS")
print(f"Valid Results:", valid)
print(f"Invalid Results:", invalid)
print("=" * 40)
print(f"DIAGNOSTIC REPORT")
print(f"Calculated Result: {round((average_invalid / average_valid), 2)}")
print(f"System State: {result}")

```

```python
## APPENDIX 2:  Recursive Fault Trace
import random
random.seed(SEED_NUM)
# Generate a random fault code based on student information.
fault_code = f"{LAST_NAME[0:2].upper()}{FAVORITE_ARTIST[0:2]}-{random.randint(1, 10)}"
parse_code = int("".join([char for char in fault_code if char.isdigit()]))

def trace_fault(n, level=1):
    print(f"Entering with {n}: at level {level}")
    # Establishing Base Case
    if n == parse_code:
        print(f"Base case reached at level {level}")
        return "Done"
    
    if n > parse_code:
        n = n // 3

    try:
        result = trace_fault(n + 1, level + 1)
    except Exception:
        print(f"Exception found at level {level}")
    print(f"Currently entering with: {n}, at level: {level}")
    return result

print(f"Generated Fault Data: {fault_code}")
print()
trace_fault(n=0)


```
